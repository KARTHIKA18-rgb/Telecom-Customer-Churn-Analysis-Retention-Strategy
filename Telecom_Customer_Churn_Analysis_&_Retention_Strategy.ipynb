# Telecom Customer Churn Analysis & Retention Strategy

### *Project Summary*

Customer churn is one of the biggest challenges faced by telecommunication companies. When customers discontinue their services and move to competitors, companies lose recurring revenue, increase customer acquisition costs, and experience slower business growth.

The telecom company has observed that a significant number of customers are cancelling their subscriptions every month. However, the company does not clearly understand the factors influencing customer churn.

To address this issue, the management has requested a detailed customer churn analysis. The goal is to identify the key factors responsible for customer attrition, measure churn-related business metrics, and provide actionable recommendations that can improve customer retention.

As a Data Analyst, our responsibility is to analyze the customer dataset, identify churn patterns, build an interactive dashboard, and help management make data-driven decisions.

### *Project Objectives*

* To analyze customer demographics, account details, and service usage to understand customer behavior.
* To identify the key factors that contribute to customer churn.
* To calculate and analyze the overall customer churn rate.
* To perform Exploratory Data Analysis (EDA) using statistical methods and visualizations.
* To examine the relationship between customer churn and factors such as contract type, tenure, monthly charges, payment method, and internet service.
* To identify high-risk customer segments that are more likely to leave the company.
* To analyze the impact of different telecom services (Phone Service, Internet Service, Tech Support, Online Security, Streaming Services, etc.) on customer retention.
* To estimate the revenue impact caused by customer churn.
* To create interactive dashboards for monitoring churn trends and customer insights.
* To generate data-driven business insights for improving customer satisfaction.
* To recommend effective customer retention strategies based on analytical findings.
* To support data-driven decision-making that helps reduce customer churn and increase customer loyalty.


# *Dataset Description*


The **Telco Customer Churn** dataset contains information about customers of a telecommunications company, including their demographic details, account information, subscribed services, billing details, and churn status. Each record represents a single customer and indicates whether the customer has stayed with or left the company. The dataset is used to analyze customer behavior, identify the factors influencing churn, and develop strategies to improve customer retention. It includes both categorical and numerical features, making it suitable for exploratory data analysis, visualization, and predictive analytics.


# The dataset includes the following key attributes:

**Churn** : Target variable.

**Contract** : Customers with month-to-month contracts are more likely to churn.

**tenure** : Short-tenure customers generally have higher churn.

**MonthlyCharges** : Higher monthly charges often correlate with higher churn.

**TotalCharges** : Indicates customer's lifetime value.

**InternetService** : Fiber optic customers often show different churn patterns.

**TechSupport** : Customers without tech support tend to churn more.

**OnlineSecurity** : Lack of online security is associated with higher churn.

**PaymentMethod** : Electronic check users often have higher churn.

**PaperlessBilling** : May influence churn behavior.

#Data Preprocessing / Feature Engineering

```python
df.info()
```

```python
# df = df.drop('customerID', axis=1)
```

```python
from sklearn.preprocessing import LabelEncoder

customerID_encoder = LabelEncoder()
gender_encoder = LabelEncoder()
partner_encoder = LabelEncoder()
Dependents_encoder = LabelEncoder()
PhoneService_encoder = LabelEncoder()
MultipleLines_encoder = LabelEncoder()
InternetService_encoder = LabelEncoder()
OnlineSecurity_encoder = LabelEncoder()
OnlineBackup_encoder = LabelEncoder()
DeviceProtection_encoder = LabelEncoder()
TechSupport_encoder = LabelEncoder()
StreamingTV_encoder = LabelEncoder()
StreamingMovies_encoder = LabelEncoder()
Contract_encoder = LabelEncoder()
PaperlessBilling_encoder = LabelEncoder()
PaymentMethod_encoder = LabelEncoder()
TotalCharges_encoder = LabelEncoder()
Churn_encoder = LabelEncoder()


df['customerID'] = customerID_encoder.fit_transform(df['customerID'])
df['gender'] = gender_encoder.fit_transform(df['gender'])
df['Partner'] = partner_encoder.fit_transform(df['Partner'])
df['Dependents'] = Dependents_encoder.fit_transform(df['Dependents'])
df['PhoneService'] = PhoneService_encoder.fit_transform(df['PhoneService'])
df['MultipleLines'] = MultipleLines_encoder.fit_transform(df['MultipleLines'])
df['InternetService'] = InternetService_encoder.fit_transform(df['InternetService'])
df['OnlineSecurity'] = OnlineSecurity_encoder.fit_transform(df['OnlineSecurity'])
df['OnlineBackup'] = OnlineBackup_encoder.fit_transform(df['OnlineBackup'])
df['DeviceProtection'] = DeviceProtection_encoder.fit_transform(df['DeviceProtection'])
df['TechSupport'] = TechSupport_encoder.fit_transform(df['TechSupport'])
df['StreamingTV'] = StreamingTV_encoder.fit_transform(df['StreamingTV'])
df['StreamingMovies'] = StreamingMovies_encoder.fit_transform(df['StreamingMovies'])
df['Contract'] = Contract_encoder.fit_transform(df['Contract'])
df['PaperlessBilling'] = PaperlessBilling_encoder.fit_transform(df['PaperlessBilling'])
df['PaymentMethod'] = PaymentMethod_encoder.fit_transform(df['PaymentMethod'])
df['TotalCharges'] = TotalCharges_encoder.fit_transform(df['TotalCharges'])
df['Churn'] = Churn_encoder.fit_transform(df['Churn'])
```
